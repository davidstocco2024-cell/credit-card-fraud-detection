
## 05. Target Variable Analysis

Analyze the target variable (Class) to understand the distribution of legitimate and fraudulent transactions.

This section evaluates:

- Target class identification.
- Absolute transaction counts.
- Relative class distribution.
- Fraud prevalence.
- Class imbalance ratio.

The results will help characterize the classification problem and establish a baseline for future model evaluation.


1. Target Class Identification

```python

target_classes = df["Class"].unique()

print(f"Unique target classes: {sorted(target_classes)}")
```

2. Class Distribution

```python

class_counts = df["Class"].value_counts().sort_index()

print("Transaction counts by class:")
print(class_counts)
```

3. Class Distribution Percentage

```python

class_percentages = (
    df["Class"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)

class_distribution = pd.DataFrame({
    "Transaction_Count": class_counts,
    "Percentage": class_percentages
})

class_distribution.index.name = "Class"

class_distribution
```

4 — Fraud Prevalence

```python

total_transactions = len(df)

fraud_transactions = (df["Class"] == 1).sum()

fraud_prevalence = (
    fraud_transactions / total_transactions
) * 100

print(f"Total transactions: {total_transactions:,}")
print(f"Fraudulent transactions: {fraud_transactions:,}")
print(f"Fraud prevalence: {fraud_prevalence:.4f}%")
```

5 — Class Imbalance Ratio

```python

legitimate_count = class_counts.loc[0]
fraud_count = class_counts.loc[1]

imbalance_ratio = legitimate_count / fraud_count

print(f"Legitimate transactions: {legitimate_count:,}")
print(f"Fraudulent transactions: {fraud_count:,}")
print(f"Class imbalance ratio: {imbalance_ratio:.2f}:1")
```


### Key Findings

- The dataset contains 284,807 transactions.
- Legitimate transactions represent approximately 99.83% of the dataset.
- Fraudulent transactions represent approximately 0.17%.
- Only 492 transactions are labeled as fraudulent.
- The class imbalance ratio is approximately 577.88:1.

**Analytical Implication:**

The dataset presents a severe class imbalance problem.

Accuracy alone will not be sufficient to evaluate future classification models, since a model could achieve high accuracy while failing to identify fraudulent transactions.

Precision, Recall, F1-score, and Precision-Recall AUC will be considered during model evaluation.

Analyze the temporal characteristics of the credit card transaction dataset.

```python

time_min = df["Time"].min()
time_max = df["Time"].max()

time_span_seconds = time_max - time_min
time_span_hours = time_span_seconds / 3600
time_span_days = time_span_hours / 24

print(f"Minimum time: {time_min:,.0f} seconds")
print(f"Maximum time: {time_max:,.0f} seconds")
print(f"Observation period: {time_span_hours:.2f} hours")
print(f"Observation period: {time_span_days:.2f} days")
```

```python

is_chronologically_ordered = (
    df["Time"].is_monotonic_increasing
)

print(
    f"Chronological ordering: "
    f"{is_chronologically_ordered}"
)
```

3 — Unique Time Values

```python

unique_time_values = df["Time"].nunique()

total_transactions = len(df)

print(f"Total transactions: {total_transactions:,}")
print(f"Unique time values: {unique_time_values:,}")
```

4 — Repeated Time Values

```python

repeated_time_records = df["Time"].duplicated().sum()

time_value_counts = df["Time"].value_counts()

repeated_time_values = (
    time_value_counts > 1
).sum()

print(
    f"Repeated time records: "
    f"{repeated_time_records:,}"
)

print(
    f"Time values shared by multiple transactions: "
    f"{repeated_time_values:,}"
)
```

## 08. Financial Transaction Profiling

Analyze the financial characteristics of transaction amounts to identify potential differences between legitimate and fraudulent transactions.

This section examines:

- Overall transaction amount distribution.
- Transaction amounts by target class.
- Central tendency and dispersion by class.
- Financial percentiles.
- Zero-value transactions.
- High-value transaction behavior.
- Comparative financial transaction statistics.

The objective is to understand the financial behavior of transactions and identify potential patterns that may be relevant for subsequent fraud detection analysis.


1 — Overall Transaction Amount Distribution

```python

amount_profile = df["Amount"].agg(
    [
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ]
)

print("Overall Transaction Amount Profile:")
display(amount_profile)
```

2 — Transaction Amount Statistics by Class

```python

amount_by_class = (
    df.groupby("Class")["Amount"]
    .agg(
        Transaction_Count="count",
        Mean_Amount="mean",
        Median_Amount="median",
        Std_Amount="std",
        Min_Amount="min",
        Max_Amount="max"
    )
)

amount_by_class
```

3 — Financial Percentile Analysis by Class

```python

percentiles = [0.25, 0.50, 0.75, 0.90, 0.95, 0.99]

amount_percentiles = (
    df.groupby("Class")["Amount"]
    .quantile(percentiles)
    .unstack()
)

amount_percentiles.columns = [
    f"P{int(p * 100)}"
    for p in percentiles
]

amount_percentiles
```

4 — Zero-Value Transaction Analysis

```python

zero_amount_counts = (
    df.assign(Is_Zero_Amount=df["Amount"] == 0)
    .groupby("Class")["Is_Zero_Amount"]
    .agg(
        Zero_Amount_Count="sum",
        Total_Transactions="count"
    )
)

zero_amount_counts["Zero_Amount_Percentage"] = (
    zero_amount_counts["Zero_Amount_Count"]
    / zero_amount_counts["Total_Transactions"]
    * 100
)

zero_amount_counts
```

5 — High-Value Transaction Analysis

```python

high_value_threshold = df["Amount"].quantile(0.95)

high_value_flags = df["Amount"] > high_value_threshold

high_value_profile = (
    high_value_flags
    .groupby(df["Class"])
    .agg(
        High_Value_Count="sum",
        Total_Transactions="count"
    )
)

high_value_profile["High_Value_Percentage"] = (
    high_value_profile["High_Value_Count"]
    / high_value_profile["Total_Transactions"]
    * 100
)

print(
    f"High-value transaction threshold: "
    f"{high_value_threshold:.2f}"
)

display(high_value_profile)
```

6 — Comparative Financial Transaction Summary

```python

financial_summary = (
    df.groupby("Class")
    .agg(
        Transaction_Count=("Amount", "count"),
        Mean_Amount=("Amount", "mean"),
        Median_Amount=("Amount", "median"),
        Std_Amount=("Amount", "std"),
        Q75=("Amount", lambda x: x.quantile(0.75)),
        Min_Amount=("Amount", "min"),
        Max_Amount=("Amount", "max")
    )
)

financial_summary["Zero_Amount_Count"] = (
    df.groupby("Class")["Amount"]
    .apply(lambda x: (x == 0).sum())
)

financial_summary["High_Value_Count"] = (
    high_value_profile["High_Value_Count"]
)

financial_summary
```


### Key Findings

The financial transaction profiling analysis provides an overview of transaction amount characteristics across legitimate and fraudulent transactions.

The analysis examines differences in central tendency, dispersion, percentile distribution, zero-value transactions, and high-value transaction frequency.

**Analytical Considerations:**

- Differences between mean and median may indicate asymmetric transaction amount distributions.
- Standard deviation provides information about transaction amount variability.
- Percentile analysis helps identify differences in the distribution of transaction values.
- Zero-value transactions may represent specific transaction behaviors.
- High-value transaction frequency provides an additional dimension for comparing both classes.

**Important Limitation:**

Transaction amount alone cannot determine whether an operation is fraudulent.

Financial transaction profiling provides descriptive evidence that may help identify potential behavioral differences. Further analysis is required to determine whether these characteristics have predictive value.



## 09. PCA Feature Profiling

Analyze the statistical characteristics of the PCA-transformed features (V1 to V28).

This section examines:

- PCA feature identification.
- Overall PCA feature statistics.
- PCA feature distribution by target class.
- Mean differences between legitimate and fraudulent transactions.
- Standard deviation comparisons.
- Feature-level statistical differences.

The objective is to understand the characteristics of the transformed variables and identify potential differences between legitimate and fraudulent transactions.


1 — PCA Feature Identification

```python

pca_features = [
    column for column in df.columns
    if column.startswith("V")
]

print(f"Number of PCA features: {len(pca_features)}")
print("PCA features:")
print(pca_features)
```

2 — Overall PCA Descriptive Statistics

```python

pca_descriptive_statistics = (
    df[pca_features]
    .describe()
    .T
)

pca_descriptive_statistics
```

3 — PCA Feature Statistics by Class

```python

pca_statistics_by_class = (
    df.groupby("Class")[pca_features]
    .agg(["mean", "std"])
)

pca_statistics_by_class
```

4 — PCA Mean Differences

```python

pca_mean_by_class = (
    df.groupby("Class")[pca_features]
    .mean()
)

pca_mean_differences = pd.DataFrame({
    "Legitimate_Mean": pca_mean_by_class.loc[0],
    "Fraud_Mean": pca_mean_by_class.loc[1]
})

pca_mean_differences["Mean_Difference"] = (
    pca_mean_differences["Fraud_Mean"]
    - pca_mean_differences["Legitimate_Mean"]
)

pca_mean_differences["Absolute_Mean_Difference"] = (
    pca_mean_differences["Mean_Difference"].abs()
)

pca_mean_differences.sort_values(
    by="Absolute_Mean_Difference",
    ascending=False
)
```

5 — PCA Standard Deviation Comparison

```python

pca_std_by_class = (
    df.groupby("Class")[pca_features]
    .std()
)

pca_std_comparison = pd.DataFrame({
    "Legitimate_Std": pca_std_by_class.loc[0],
    "Fraud_Std": pca_std_by_class.loc[1]
})

pca_std_comparison["Std_Difference"] = (
    pca_std_comparison["Fraud_Std"]
    - pca_std_comparison["Legitimate_Std"]
)

pca_std_comparison["Absolute_Std_Difference"] = (
    pca_std_comparison["Std_Difference"].abs()
)

pca_std_comparison.sort_values(
    by="Absolute_Std_Difference",
    ascending=False
)
```

6 — PCA Feature Comparison Summary

```python

pca_feature_summary = pd.DataFrame({
    "Legitimate_Mean": pca_mean_by_class.loc[0],
    "Fraud_Mean": pca_mean_by_class.loc[1],
    "Mean_Difference": pca_mean_differences["Mean_Difference"],
    "Absolute_Mean_Difference": pca_mean_differences[
        "Absolute_Mean_Difference"
    ],
    "Legitimate_Std": pca_std_by_class.loc[0],
    "Fraud_Std": pca_std_by_class.loc[1],
    "Std_Difference": pca_std_comparison["Std_Difference"]
})

pca_feature_summary = pca_feature_summary.sort_values(
    by="Absolute_Mean_Difference",
    ascending=False
)

pca_feature_summary
```


### Key Findings

The PCA feature profiling analysis provides an overview of the statistical characteristics of the 28 transformed variables.

The analysis examines differences in means and standard deviations between legitimate and fraudulent transactions.

**Analytical Considerations:**

- PCA features represent transformed numerical components.
- Mean differences indicate variations in the average component values between target classes.
- Standard deviation differences provide information about class-specific variability.
- Some PCA components may exhibit more pronounced descriptive differences than others.

**Important Limitations:**

- Descriptive differences do not establish statistical significance.
- Mean differences alone do not demonstrate predictive power.
- PCA components do not provide direct information about the original financial meaning of individual variables.
- Further statistical analysis and model evaluation are required to assess their relevance for fraud detection.


### Dataset Limitations

The following limitations must be considered when interpreting the dataset and designing subsequent analytical stages.

**1. Class Imbalance**

Fraudulent transactions represent a very small proportion of the dataset, creating a highly imbalanced classification problem.

**2. Limited Temporal Information**

The Time feature represents elapsed seconds rather than actual calendar timestamps. Therefore, calendar-based behavioral patterns cannot be directly reconstructed.

**3. Anonymized Features**

The V1 to V28 variables are PCA-transformed components. Their original financial meaning and underlying source variables are not available.

**4. Limited Financial Context**

The Amount feature provides transaction values but does not include additional contextual information such as merchant identity, customer profile, transaction location, or transaction channel.

**5. Historical Dataset**

The dataset represents a historical collection of transactions. Its statistical characteristics may not fully reflect current fraud patterns or future transaction behavior.

**6. Limited Fraud Observations**

The relatively small number of fraudulent transactions may restrict the statistical analysis and generalization of future models.
